### 3.6 Analisis Outlier dengan IQR (Interquartile Range)

Sesuai materi *00.ML_Process.docx*:
* Menghitung kuartil pertama (Q1) dengan `.quantile(0.25)` dan kuartil ketiga (Q3) dengan `.quantile(0.75)`.
* Interquartile Range (IQR) dihitung dengan mengurangkan Q1 dari Q3: $IQR = Q3 - Q1$.
* Batas Bawah (*Lower Bound*) = $Q1 - (1.5 \times IQR)$.
* Batas Atas (*Upper Bound*) = $Q3 + (1.5 \times IQR)$.
* Nilai yang berada di luar rentang ini dianggap sebagai *outlier*.
* *Catatan Metode:* IQR sangat cocok untuk data yang distribusinya tidak normal dan miring.

```python
# Analisis Outlier dengan IQR
for col in ['YearsExperience', 'Salary']:
    data_col = df[col].dropna()
    q1 = data_col.quantile(0.25)
    q3 = data_col.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    outliers_iqr = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"=== Analisis Outlier IQR: {col} ===")
    print(f"Q1 (Kuartil 1) : {q1:.2f}")
    print(f"Q3 (Kuartil 3) : {q3:.2f}")
    print(f"IQR            : {iqr:.2f}")
    print(f"Batas Bawah    : {lower_bound:.2f}")
    print(f"Batas Atas     : {upper_bound:.2f}")
    print(f"Jumlah Outlier : {len(outliers_iqr)}")
    if len(outliers_iqr) > 0:
        print("Data yang termasuk Outlier:")
        display(outliers_iqr)
    print("-" * 50)
```

**Insight dari Analisis Outlier dengan IQR:**  
* **`YearsExperience`:** Batas bawah = -8.65 tahun dan batas atas = 28.55 tahun. Karena seluruh nilai karyawan berada pada rentang 0.7 - 19.9 tahun, maka **tidak ada outlier (0 outlier)**.
* **`Salary`:** Batas bawah = -Rp 94.050 dan batas atas = Rp 349.070.
* **Nilai Outlier yang Ditemukan:** Tepat **1 baris data outlier**, yaitu pada **indeks ke-59** dengan masa kerja `YearsExperience = 17.0` tahun dan nominal gaji `Salary = Rp 518.900`. Nilai ini jauh melampaui batas wajar atas Rp 349.070.

### 3.7 Analisis Outlier dengan Z-Score

Sesuai materi *00.ML_Process.docx*, metode analisis Z-Score cocok untuk data yang distribusinya normal (*bell curve*) serta data tidak miring.  
Rumus Z-Score:
$$Z = \frac{X - \mu}{\sigma}$$
Di mana $\mu$ adalah rata-rata (*mean*) dan $\sigma$ adalah standar deviasi (*standard deviation*). Data dianggap sebagai outlier jika memiliki $|Z| > 3$ (berada lebih dari 3 standar deviasi dari rata-rata).

```python
# Analisis Outlier dengan Z-Score
for col in ['YearsExperience', 'Salary']:
    data_col = df[col].dropna()
    mean = data_col.mean()
    std = data_col.std()
    
    z_scores = (df[col] - mean) / std
    outliers_z = df[z_scores.abs() > 3]
    
    print(f"=== Analisis Outlier Z-Score: {col} ===")
    print(f"Mean (Rata-rata)  : {mean:.2f}")
    print(f"Std Deviasi       : {std:.2f}")
    print(f"Jumlah Outlier (|Z| > 3): {len(outliers_z)}")
    if len(outliers_z) > 0:
        print("Data Outlier beserta Nilai Z-Score:")
        outlier_detail = outliers_z.copy()
        outlier_detail['Z_Score'] = z_scores[z_scores.abs() > 3]
        display(outlier_detail)
    print("-" * 50)
```

**Insight dari Analisis Outlier dengan Z-Score:**  
* Pada kolom `YearsExperience`, tidak ada nilai yang memiliki $|Z| > 3$ (0 outlier).
* Pada kolom `Salary`, terdeteksi **1 outlier** pada baris **indeks ke-59** dengan nilai **Z-Score = +5.12**.
* Z-Score sebesar 5.12 artinya gaji karyawan tersebut berada lebih dari 5 kali standar deviasi di atas rata-rata populasi karyawan perusahaan. Hal ini secara statistik memvalidasi bahwa titik tersebut merupakan anomali ekstrem.
